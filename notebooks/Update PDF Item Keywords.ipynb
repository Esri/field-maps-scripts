#### Run this cell to connect to your GIS and get started:

```python
from arcgis.gis import GIS
gis = GIS("home")
```

#### Choose the keyword to add

```python
#keyword_to_add = "FieldMapsDisabled"
keyword_to_add = "Geospatial PDF"
```

#### Find PDFs owned by the current user

```python
username = gis.users.me.username
query = 'type:"PDF" AND owner:{}'.format(username)
pdf_items = gis.content.search(query=query, max_items=1000)

print(f"Found {len(pdf_items)} PDF items")
```

#### Or Find all PDFs in a group

```python
GROUP_ID = "2905d09e8828444f9dff3fc702f620ed"
group = gis.groups.get(GROUP_ID)
pdf_items = [item for item in group.content(max_items=1000) if item.type == "PDF"]

print(f"Found {len(pdf_items)} PDF items")
```

#### Add the keyword

```python
if not keyword_to_add:
    raise ValueError("Select a keyword to add.")

updated = 0
skipped = 0
failed = 0

for item in pdf_items:
    try:
        current = set(item.typeKeywords or [])

        if keyword_to_add in current:
            skipped += 1
            print(f"Skipped '{item.title}' - keyword already present")
            continue

        new_type_keywords = sorted(current | {keyword_to_add})
        result = item.update({"typeKeywords": new_type_keywords})
        result = True

        if result:
            updated += 1
            print(f"Updated '{item.title}' - added: {keyword_to_add}")
        else:
            failed += 1
            print(f"Failed to update '{item.title}'")

    except Exception as e:
        failed += 1
        print(f"Failed on '{item.title}': {e}")

print(f"Done. Updated={updated}, Skipped={skipped}, Failed={failed}")
```
